# Tokenize

```python
from transformers import AutoTokenizer
```

```python
text = df['text'][0]
print(f"Text: \n\n {text[:1000]}", end="\n\n")

tokenizer = AutoTokenizer.from_pretrained("GSAI-ML/LLaDA-8B-Instruct")
tokens = tokenizer(text)["input_ids"]
print(f"Token length: {len(tokens)}\n\n")
print(f"Tokens: {tokens[:100]}\n\n")
```

# Chunking

```python
curr_chunks = []
curr_chunks.extend(tokens)

print(f"Curr chunks: {curr_chunks[:100]}")
```

# Making Tokens

```python
_max = 4096

if len(curr_chunks) < _max:
    print("Smaller than max")
    
    while len(curr_chunks) < _max:
        curr_chunks.append(0)

curr_tokens = curr_chunks[:_max]

print(f"Curr tokens: {curr_tokens[:100]}")
print(f"Curr tokens length: {len(curr_tokens)}")
```

# Diff Masking

```python
import torch
from random import random
```

```python
_id_mask_token = 126336
eps = 1e-3

chunk_tensor  = torch.tensor(curr_tokens, dtype=torch.int32)

t = random()
p_mask = (1.0 - eps) * t + eps

mask = torch.rand(chunk_tensor.size(0)) < p_mask
noisy = chunk_tensor.clone()
noisy[mask] = _id_mask_token

print(f"Original: {chunk_tensor[:100]}")
print(f"Noisy: {noisy[:100]}")
print(f"Mask: {mask[:100]}")
print(f"p_mask: {p_mask}, t: {t}")
print(f"Number of masked tokens: {mask.sum().item()}")
```

# Viz

```python
def decode(ids):
    return tokenizer.decode(ids, skip_special_tokens=False)

print(f"Original decoded: {decode(chunk_tensor[:100])} \n\n")
print(f"Noisy decoded: {decode(noisy[:100])}\n\n")
```

# Training data Structure

```python
training_data = {
    "t": t,
    "input_ids": chunk_tensor,
    "noisy_input_ids": noisy,
    "mask"  : mask,
}

for k, v in training_data.items():
    if isinstance(v, torch.Tensor):
        print(k, v.shape, v.dtype)
        print()
        print(f"{k}: {v[:100]}")
    else:
        print(k, v)
        
```
